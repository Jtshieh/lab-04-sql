# Mockaroo data inspection

The CSV contains 200 synthetic records generated with Mockaroo. The `group` field randomly selects blue, gold, green, or purple. Its blank probability is 5%, as are those of `email` and `ip_address`. Other fields have no requested blanks. The file uses a header and Unix line endings.

```python
import pandas as pd
from IPython.display import display

df = pd.read_csv("MOCK_DATA.csv")
display(df.head())
```

```python
display(df.dtypes.rename("pandas dtype").to_frame())
display(df.isna().sum().rename("Missing rows").to_frame())
```

## SQL types

`id` is an integer and maps to `BIGINT`, with a primary key. The remaining columns contain strings and map to `VARCHAR(255)`. `group` is quoted with backticks in SQL because it is a reserved word. Rows with missing values are removed before upload, so all text columns use `NOT NULL`.

```python
cleaned = df.copy()
for column in cleaned.columns[1:]:
    cleaned[column] = cleaned[column].astype("string").str.strip().replace("", pd.NA)
cleaned = cleaned.dropna()
print(f"Original rows: {len(df)}")
print(f"Complete rows: {len(cleaned)}")
display(cleaned["group"].value_counts().sort_index().rename("Rows").to_frame())
```
